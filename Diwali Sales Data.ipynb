```python
import pandas as pd
```

```python
df=pd.read_csv("Diwali Sales Data.csv",encoding="unicode_escape")
```

```python
df
```

```python
# 1. Delete Cols - 'Status', 'Unnamed':-
df=df.drop(['Status','unnamed1'],axis=1)
df
```

```python
# 2. Check null values (Delete if less than 5%):-
#Checking null values:
print(df.isnull().sum())
```

```python
#checking null values %:
print(df.isnull().mean()*100)
```

```python
# Deleting null values (less than 5%):
df=df.dropna(subset=["Amount"])
```

```python
#Checking for null values again:
print(df.isnull().sum())
```

```python
# 3. Rename 'Marital_Status' to 'Shaadi':
df=df.rename(columns={"Marital_Status":"Shaadi"})
df
```

```python
# 4. Age Group wise Orders:
age_group_orders=df.groupby("Age Group")["Orders"].sum()
age_group_orders
```

```python
# 5. Count of Male & Female:
Gender_Count=df["Gender"].value_counts()
Gender_Count
```

```python
# 6. State wise Amount:
State_Amt=df.groupby("State")["Amount"].sum().sort_values(ascending=False)
State_Amt
```

```python
# 7. Occupation wise Orders:
Occ_order=df.groupby("Occupation")["Orders"].sum().sort_values(ascending=False)
Occ_order
```

```python
# 8. Product ID wise Amounts/Orders:
#Product ID wise Amounts:
ProductID_Amt=df.groupby("Product_ID")["Amount"].sum()
ProductID_Amt
```

```python
#Product ID wise Orders:
ProductID_Order=df.groupby("Product_ID")["Orders"].sum()
ProductID_Order
```

```python
group=df.groupby("Zone")["User_ID"].count()
print(group)
```

# Exploratory Data Analysis (EDA) :~

### Diwali Sales Data

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Gender :

```python
# Ploting, a bar chart for gender and it's count :

C=['#FF99FF','#99CCFF']
a=sns.countplot(x='Gender',data=df,hue="Gender",palette=C)

for bars in a.containers:
    a.bar_label(bars)
```

```python
# Ploting a bar chart for gender vs total amount :

sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sales_gen
sns.barplot(x='Gender',y='Amount',data=sales_gen, hue='Amount')
plt.title("Gender vs Total Amount")
plt.show()
```

#### Age :

```python
age=sns.countplot(x='Age Group',data=df,hue="Gender")
for bars in age.containers:
    age.bar_label(bars)
```

```python
# Ploting a bar chart for total amount vs age group :

sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sales_age

sns.barplot(x='Age Group',y='Amount',data=sales_age,hue='Amount')
plt.title("Total Amount vs Age Group")
plt.show()
```

#### State :

```python
# Total number of orders from top 10 states    (Descending):

State_order=df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
State_order

sns.barplot(x='State', y='Orders',data=State_order,hue='Orders')
plt.title('Total number of orders from top 10 states (Descending)')
plt.xticks(rotation=90)
plt.show()
```

```python
# Total number of orders from top 10 states    (Ascending):

state_Order=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=True).head(10)
state_Order

sns.set(rc={'figure.figsize':(18,5)})
sns.barplot(x='State', y='Orders', data=state_Order, hue='Orders')
plt.title('Total number of orders from top 10 states (Ascending)')
plt.show()
```

```python
# Total amount/sales from top 10 states:

amt_state=df.groupby(['State'],as_index=False) ['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
amt_state

sns.barplot(x='State',y='Amount',data=amt_state,hue='Amount')
plt.title('Total amount/sales from top 10 states')
plt.show()
```

#### Marital Status :

```python
m=sns.countplot(x='Shaadi',data=df, hue='Shaadi')
sns.set(rc={'figure.figsize':(10,5)})
plt.title('Marital Status')
plt.show()

for bars in m.containers:
    m.bar_label(bars)
```

```python
# Ploting a bar chart for marital status vs total amount :

m_amt=df.groupby(['Shaadi','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
m_amt

sns.set(rc={'figure.figsize':(10,5)})
sns.barplot(data=m_amt, x='Shaadi', y='Amount',hue='Gender')
plt.title('marital status of each gender vs total amount')
plt.show()
```

#### Occupation :

```python
occ=sns.countplot(x='Occupation',data=df,hue='Occupation')
plt.xticks(rotation=90)
plt.title("Occupation")

for bars in occ.containers:
    occ.bar_label(bars)
```

```python
# Ploting a bar chart for Occupation vs total amount :

occ_amt=df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Occupation', y='Amount',data=occ_amt, hue='Amount')
sns.set(rc={'figure.figsize':(20,5)})
plt.title('Occupation vs total amount')
plt.show()
```

#### Product Category :

```python
pc=sns.countplot(x='Product_Category',data=df,hue='Product_Category')
plt.xticks(rotation=90)

plt.title('Product Category')
for bars in pc.containers:
    pc.bar_label(bars)
```

```python
# Ploting a bar chart for Product category vs total amount :

occ_amt=df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Product_Category', y='Amount',data=occ_amt, hue='Amount')
plt.xticks(rotation=90)
plt.title('Product category vs total amount ')
plt.show()
```

```python
# Ploting a bar chart for Product category vs Orders (Top 10):

occ_amt=df.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.barplot(x='Product_ID', y='Orders',data=occ_amt, hue='Orders')
plt.xticks(rotation=90)
plt.title('Product ID vs Orders (Top 10)')
plt.show()
```
